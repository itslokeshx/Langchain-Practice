```python
from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_ollama import OllamaEmbeddings
from langchain_community.vectorstores import FAISS

loader=TextLoader("../data/speech.txt")
documents=loader.load()
text_splitter=RecursiveCharacterTextSplitter(chunk_size=500,chunk_overlap=30)
docs=text_splitter.split_documents(documents)
```

```python
docs
```

```python
embeddings=OllamaEmbeddings(model="nomic-embed-text")  
db_new=FAISS.from_documents(docs,embeddings)
db_new
```

```python
### querying 
query="Does Kohli exercise?"
docs=db_new.similarity_search(query)
docs
```

```python
retriever=db_new.as_retriever()
docs=retriever.invoke(query)
docs[0].page_content
```

```python
docs_and_score=db_new.similarity_search_with_score(query)
docs_and_score
```

```python
embedding_vector=embeddings.embed_query(query)
embedding_vector
```

```python
docs_score=db_new.similarity_search_by_vector(embedding_vector)
docs_score
```

```python
### Saving And Loading
db_new.save_local("faiss_index")
```

```python
new_db=FAISS.load_local("faiss_index",embeddings,allow_dangerous_deserialization=True)
docs=new_db.similarity_search(query)
```
